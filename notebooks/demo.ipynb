# Jingjiang-MCMM-GAT: Quick Demonstration

This notebook demonstrates how to load static revetment features (`is_protected`), initialize the `MigrationGAT` model, and run inference on the Jingjiang Reach 1000-node graph.


```python
import torch
import pandas as pd
import numpy as np
from pathlib import Path
import sys

# Add src to python path
sys.path.insert(0, str(Path("..").resolve() / "src" / "gnn"))
from models.gat_model import MigrationGATSimple

print("PyTorch version:", torch.__version__)
```

```python
# 1. Load static engineering protection features (is_protected)
eng_csv = Path("../data/processed/engineering_features.csv")
df_eng = pd.read_csv(eng_csv)
print(f"Loaded {len(df_eng)} reach nodes.")
n_prot = df_eng["is_protected"].sum()
print(f"Protected nodes (is_protected=1): {n_prot} / {len(df_eng)}")
df_eng.head()
```

```python
# 2. Instantiate MigrationGAT model
# 15 static features + 7 hydrological features = 22 input dims
model = MigrationGATSimple(
    in_channels=22,
    hidden_channels=32,
    out_channels=1,
    heads=2,
    num_static=15,
    num_hydro=7
)
print("Model architecture:")
print(model)
```

```python
# 3. Check trained model checkpoint if available
ckpt_path = Path("../outputs/cv_monthly_results/fold1_model.pt")
if ckpt_path.exists():
    checkpoint = torch.load(ckpt_path, map_location="cpu")
    model.load_state_dict(checkpoint)
    print("Loaded trained GAT model checkpoint from:", ckpt_path)
else:
    print("No checkpoint found at", ckpt_path, "- run Step 3 in README to train.")
```
